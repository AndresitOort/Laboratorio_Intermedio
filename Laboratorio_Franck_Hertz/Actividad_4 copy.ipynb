```python
import os
import numpy as np
import pandas as pd
import glob
import os
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

plt.style.use('bmh')
```

```python
ruta = "/Users/andresitoortiz/Library/Mobile Documents/com~apple~CloudDocs/Pregrado_Fisica/2025-2/Lab_intermedio/Franck-Hertz/FranckHertz/Actividad_4"
data = {}

for archivo in glob.glob(os.path.join(ruta, "*.csv")):
    nombre = os.path.basename(archivo)
    df = pd.read_csv(archivo, delimiter=';')
    df = df.replace(',', '.', regex=True)
    data[nombre] = df

print("Archivos cargados:", list(data.keys()))
voltajes = [6.3,6.8,6.7,6.0]
print(data.values())
```

```python
solarized_colors = [
    "#268BD2",  # azul
    "#2AA198",  # verde-azulado
    "#859900",  # verde amarillento
    "#CB4B16",  # naranja profundo
    "#DC322F",  # rojo
    "#B58900",  # amarillo tostado
    "#6C71C4",  # púrpura
    "#586E75"   # gris azulado
]
```

```python
fig,axes = plt.subplots(1,figsize=(12,8))
axes.set_facecolor("#FFFFFF")
i=0

for df in data.values():
    dff = df[2:-1].astype(float)
    x = dff["Voltage U1"]
    y = dff["Corriente IA"]
    
    if i==0: #6.3V
        axes.errorbar(x[2:1289:40], y[2:1289:40],
            xerr=0.01, yerr=0.01,
            fmt='o',color = solarized_colors[i] , markersize = 5, markeredgewidth = 0.5, markeredgecolor = 'k',
            capsize = 2, elinewidth = 1, ecolor = 'k',
            label=fr'$U_H = {voltajes[i]} V$')
        axes.plot(x[2:1289], y[2:1289], color = solarized_colors[i], linewidth = 1)
           
        
    #elif i==1: #6.8V
        
        #axes.errorbar(x[2:-1:50], y[2:-1:50],
            #xerr=0.01, yerr=0.01,
            #fmt='o',color = solarized_colors[i] , markersize = 5, markeredgewidth = 0.5, markeredgecolor = 'k',
            #capsize = 2, elinewidth = 1, ecolor = 'k',
            #label=fr'$U_H = {voltajes[i]} V$')
        #axes.plot(x[2:-1], y[2:-1], color = solarized_colors[i], linewidth = 1)
    
    elif i==2: #6.0V
        axes.errorbar(x[2:1098:40], y[2:1098:40],
            xerr=0.01, yerr=0.01,
            fmt='o',color = solarized_colors[i] , markersize = 5, markeredgewidth = 0.5, markeredgecolor = 'k',
            capsize = 2, elinewidth = 1, ecolor = 'k',
            label=fr'$U_H = {voltajes[i]} V$')
        axes.plot(x[2:1098], y[2:1098], color = solarized_colors[i], linewidth = 1)
    
    elif i==3: #6.0V
        axes.errorbar(x[2:1490:40], y[2:1490:40],
            xerr=0.01, yerr=0.01,
            fmt='o',color = solarized_colors[i] , markersize = 5, markeredgewidth = 0.5, markeredgecolor = 'k',
            capsize = 2, elinewidth = 1, ecolor = 'k',
            label=fr'$U_H = {voltajes[i]} V$')
        axes.plot(x[2:1490], y[2:1490], color = solarized_colors[i], linewidth = 1)
        
    i+=1
axes.set_title("Curva de Franck-Hertz para Distintos Voltajes de Termoionización " + fr"$U_H$", fontsize=20)
axes.set_xlabel(fr"$U_1 (V)$", fontsize=20)
axes.set_ylabel(fr"$I_A (nA)$", fontsize=20)
plt.legend(fontsize=16)
plt.grid(False)
plt.savefig("Curvas_Franck_Hertz_Voltajes_Termoionizacion.png", dpi=300)  
plt.show()
```
